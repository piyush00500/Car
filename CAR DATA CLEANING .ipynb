##### Location wise total sales (Price)

```python
x = df.groupby('Location')['Price'].sum().sort_values().index
y = df.groupby('Location')['Price'].sum().sort_values().values

plt.figure(figsize=(10, 6))

ax = sns.barplot(x=x, y=y, color='green')

# Add bar labels
ax.bar_label(ax.containers[0], fmt='%.0f', padding=3)

plt.xticks(rotation=45)
plt.yticks(range(0, 11000, 1000))
plt.xlabel('Location')
plt.ylabel('Total Price')
plt.title('Total Price by Location')

plt.tight_layout()
plt.show()
```

```python
x = df.groupby('Location')['Price'].sum().sort_values().index
y = df.groupby('Location')['Price'].sum().sort_values().values

ax = sns.barplot(x=x, y=y, color='orange')

ax.bar_label(ax.containers[0], fmt='%.0f', padding=3)

plt.xticks(rotation=45)
plt.yticks(list(np.arange(0, 11000, 1000)))
plt.show()
```

# Multivariate Analysis

```python
df.head()
```

```python
# Location and Fuel_Type wise total price
ax = df.groupby(['Location', 'Fuel_Type'])['Price'].sum() \
       .sort_values(ascending=False) \
       .plot(kind='bar', figsize=(12, 8), color='skyblue')

ax.bar_label(ax.containers[0], fmt='%.0f', padding=3)

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
Loca_fuel = df.groupby(['Location','Fuel_Type'])['Price'].sum().reset_index()
```

```python
plt.figure(figsize=(12, 8))

ax = sns.barplot(
    x='Location',
    y='Price',
    hue='Fuel_Type',
    data=Loca_fuel,
    palette=['skyblue', 'green', 'orange', 'pink']  # change colors as needed
)

# Add bar labels
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=3, fontsize=9)

plt.xticks(rotation=45)
plt.yticks(list(np.arange(0, 9000, 1000)))
plt.tight_layout()
plt.show()
```

```python
# Fuel type and transmission wise avg sales

df.columns
```

```python
Loca_trans = df.groupby(['Transmission', 'Fuel_Type'])['Price'].mean().reset_index()

ax = sns.barplot(
    x='Fuel_Type',
    y='Price',
    hue='Transmission',
    data=Loca_trans,
    palette=['skyblue', 'green']
)

# Add bar labels
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
Loca_trans = df.groupby(['Transmission','Fuel_Type'])['Price'].mean().reset_index()
sns.scatterplot(x='Fuel_Type',y='Price',hue='Transmission',data=Loca_trans)
plt.xticks(rotation = 45)
plt.show()
```

```python
df.head()
```

```python
# Company name and transmission wise avg price

plt.figure(figsize=(12,8))
comp_trans = df.groupby(['Company_name','Transmission'])['Price'].mean().reset_index()

sns.scatterplot(x='Company_name',y='Price',hue='Transmission',data=comp_trans)
plt.xticks(rotation = 90)
plt.grid()
plt.show()
```
